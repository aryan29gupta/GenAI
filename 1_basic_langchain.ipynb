## Connect with gemini model

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0,
    timeout=30,
    max_retries=1
)
```

```python
res = llm.invoke("Who is the PM of India? and america and what is the capital of India? and america")
```

```python
print(res.content[0]["text"])
```
